```python
import ast
# ast.literal_eval()
```

```python
def convert(obj):
  L = []
  for i in ast.literal_eval(obj):
      L.append(i["name"])
  return L
```

```python
df["genres"] = df["genres"].apply(convert)
```

```python
df.head()
```

```python
df["keywords"] = df["keywords"].apply(convert)
```

```python
def extract_cast(obj):
    count = 0
    L = []
    for i in ast.literal_eval(obj):
        if count >= 3:
            break
        L.append(i["character"])
        count += 1
    return L
```

```python
df["cast"] = df["cast"].apply(extract_cast)
```

```python
df
```

```python
df["crew"][0]
```

```python
def extract_crew(obj):
    L = []
    for i in ast.literal_eval(obj):
        if i["job"] == "Director":
            L.append(i["name"])
    return L
```

```python
df["crew"] = df["crew"].apply(extract_crew)
```

```python
final_df  = df[["id","title","tags"]]
```

```python
final_df
```

```python
final_df["tags"] = final_df["tags"].apply(lambda x:" ".join(x))
```

```python
final_df["tags"][2]
```

```python
final_df["tags"] = final_df["tags"].apply(lambda x:x.lower())
```

```python
final_df
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def extract_word(txt):
  L = []
  for i in txt.split():
    L.append(ps.stem(i))

  return " ".join(L)
```

```python
final_df["tags"] = final_df["tags"].apply(extract_word)
```

```python
final_df.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(max_features=5000,stop_words="english")
```

```python
vectors = cv.fit_transform(final_df["tags"]).toarray()
```

```python
vectors
```

```python
cv.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
cs = cosine_similarity(vectors)
```

```python
cs.shape
```

```python
sorted(list(enumerate(cs[4])),reverse=True,key=lambda x:x[1] )[1:6]
```

```python
def recommend(movie):
  movie_idx = final_df[final_df["title"] == movie].index[0]
  movie_list = sorted(list(enumerate(cs[movie_idx])),reverse=True,key=lambda x:x[1] )[1:6]

  for i in movie_list:
    print([i[0]])
  return


```

```python
recommend("Æon Flux")
```
